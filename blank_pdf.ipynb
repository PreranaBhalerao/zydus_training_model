````python
import json, random, re
from pathlib import Path
from collections import defaultdict

if "FastLanguageModel" not in dir():
    from unsloth import FastLanguageModel
if "model" not in dir() or "tokenizer" not in dir():
    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name="unsloth/Qwen2.5-1.5B-Instruct",
        max_seq_length=8192,
        dtype=None,
        load_in_4bit=True,
    )
    model.load_adapter("/content/drive/MyDrive/invoice_lora")
FastLanguageModel.for_inference(model)

FIELDS = [
    "invoice_no", "invoice_date", "vendor", "customer",
    "vendor_gstin", "customer_gstin", "tax", "total", "irn",
]
KEY_FIELDS = ["invoice_no", "vendor_gstin", "customer_gstin", "total", "tax"]
MONEY = {"tax", "total"}

def norm(value):
    if value is None:
        return ""
    text = re.sub(r"\s+", " ", str(value)).strip().upper().replace(",", "")
    return "" if text in {"", "NONE", "NULL"} else text

def same(key, pred, gold):
    a, b = norm(pred), norm(gold)
    if key in MONEY and a and b:
        try:
            return abs(float(a) - float(b)) < 0.01
        except ValueError:
            return False
    return a == b

val_path = Path("/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl")
by_vendor = defaultdict(list)
with val_path.open(encoding="utf-8") as fh:
    for line in fh:
        row = json.loads(line)
        by_vendor[(row.get("vendor") or "UNKNOWN").strip().upper()].append(row)

rng = random.Random(7)
sample = []
for vendor in rng.sample(list(by_vendor), k=len(by_vendor)):
    picks = by_vendor[vendor][:]
    rng.shuffle(picks)
    sample.extend(picks[:2])
    if len(sample) >= 100:
        break
sample = sample[:100]

stats = {k: {"match": 0, "missing": 0, "wrong": 0, "gold_present": 0} for k in FIELDS}
stats["line_count"] = {"match": 0, "missing": 0, "wrong": 0, "gold_present": 0}
scored = 0
key_pass = 0

for row in sample:
    gold = row["gemini_json"]["invoice_summary"]
    gold_items = (row["gemini_json"].get("line_items") or {}).get("items") or []
    prompt = INSTRUCTION + "\n\nOCR text:\n" + row["ocr_text"].strip()
    inputs = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
    ).to(model.device)
    if inputs.shape[-1] > 6000:
        continue
    out = model.generate(
        input_ids=inputs,
        attention_mask=inputs.ne(tokenizer.pad_token_id),
        max_new_tokens=2048,
        do_sample=False,
        use_cache=True,
    )
    text = tokenizer.decode(out[0][inputs.shape[-1]:], skip_special_tokens=True)
    summary, items = {}, []
    try:
        pred = json.loads(text[text.find("{") : text.rfind("}") + 1])
        summary = pred.get("invoice_summary") or {}
        items = (pred.get("line_items") or {}).get("items") or []
    except Exception:
        pass
    invoice_ok = True
    for key in FIELDS:
        g = gold.get(key)
        if not norm(g):
            continue
        stats[key]["gold_present"] += 1
        p = summary.get(key)
        if not norm(p):
            stats[key]["missing"] += 1
            ok = False
        elif same(key, p, g):
            stats[key]["match"] += 1
            ok = True
        else:
            stats[key]["wrong"] += 1
            ok = False
        if key in KEY_FIELDS and not ok:
            invoice_ok = False
    stats["line_count"]["gold_present"] += 1
    if not items:
        stats["line_count"]["missing"] += 1
    elif len(items) == len(gold_items):
        stats["line_count"]["match"] += 1
    else:
        stats["line_count"]["wrong"] += 1
    if invoice_ok:
        key_pass += 1
    scored += 1
    if scored % 10 == 0:
        print("scored", scored, flush=True)

total_match = sum(s["match"] for k, s in stats.items() if k != "line_count")
total_gold = sum(s["gold_present"] for k, s in stats.items() if k != "line_count") or 1
print("\nscored invoices", scored)
print(f"overall field accuracy {total_match / total_gold:.1%}")
print(f"invoices with invoice_no, both GSTINs, tax, and total all correct {key_pass / scored:.1%}")
print(f"\n{'field':<18} {'accuracy':>8} {'missing':>8} {'wrong':>7}")
rows = []
for key, s in stats.items():
    n = s["gold_present"] or 1
    rows.append((s["match"] / n, key, s))
for _, key, s in sorted(rows):
    n = s["gold_present"] or 1
    print(f"{key:<18} {s['match']/n:8.1%} {s['missing']/n:8.1%} {s['wrong']/n:7.1%}")
````

````python
import json, random, re
from pathlib import Path
from collections import defaultdict

from unsloth import FastLanguageModel

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="/content/drive/MyDrive/invoice_lora_v3",
    max_seq_length=8192,
    dtype=None,
    load_in_4bit=True,
)
FastLanguageModel.for_inference(model)

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)

FIELDS = [
    "invoice_no", "invoice_date", "vendor", "customer",
    "vendor_gstin", "customer_gstin", "tax", "total", "irn",
]

def get_summary(obj):
    if not isinstance(obj, dict):
        return {}
    return obj.get("invoice_summary") or {}

def line_count(obj):
    if not isinstance(obj, dict):
        return None
    li = obj.get("line_items")
    if isinstance(li, dict) and isinstance(li.get("items"), list):
        return len(li["items"])
    if isinstance(li, list):
        return len(li)
    return None

def norm(v):
    if v is None:
        return ""
    return re.sub(r"\s+", " ", str(v).strip().upper())

def parse_model(text):
    text = text.strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None

# Same sample style: 2 invoices per held-out vendor, skip long prompts
by_vendor = defaultdict(list)
with Path("/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl").open(encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)
        by_vendor[row.get("vendor") or "UNKNOWN"].append(row)

random.seed(3407)
sample = []
for vendor, rows in sorted(by_vendor.items(), key=lambda x: -len(x[1])):
    if len(sample) >= 24:
        break
    random.shuffle(rows)
    for row in rows[:2]:
        sample.append(row)

stats = {k: {"ok": 0, "missing": 0, "wrong": 0, "n": 0} for k in FIELDS + ["line_count"]}
strict_ok = 0
scored = 0

for i, row in enumerate(sample, 1):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + row["ocr_text"].strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    if len(tokenizer(text, add_special_tokens=False)["input_ids"]) > 6000:
        continue
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=2048, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    pred = parse_model(raw)
    gold = row["gemini_json"]
    gs, ps = get_summary(gold), get_summary(pred) if pred else {}

    for field in FIELDS:
        g, p = norm(gs.get(field)), norm(ps.get(field)) if pred else ""
        stats[field]["n"] += 1
        if not p:
            stats[field]["missing"] += 1
        elif p == g:
            stats[field]["ok"] += 1
        else:
            stats[field]["wrong"] += 1

    gc, pc = line_count(gold), line_count(pred) if pred else None
    stats["line_count"]["n"] += 1
    if pc is None:
        stats["line_count"]["missing"] += 1
    elif pc == gc:
        stats["line_count"]["ok"] += 1
    else:
        stats["line_count"]["wrong"] += 1

    need = ["invoice_no", "vendor_gstin", "customer_gstin", "tax", "total"]
    if pred and all(norm(ps.get(k)) == norm(gs.get(k)) and norm(ps.get(k)) for k in need):
        strict_ok += 1
    scored += 1
    if scored % 10 == 0:
        print("scored", scored)

total_ok = sum(stats[k]["ok"] for k in stats)
total_n = sum(stats[k]["n"] for k in stats)
print("\nscored invoices", scored)
print(f"overall field accuracy {100 * total_ok / max(total_n, 1):.1f}%")
print(f"invoices with invoice_no, both GSTINs, tax, and total all correct {100 * strict_ok / max(scored, 1):.1f}%")
print(f"\n{'field':18} {'accuracy':>8} {'missing':>8} {'wrong':>8}")
for field in sorted(stats, key=lambda k: stats[k]["ok"] / max(stats[k]["n"], 1)):
    s = stats[field]
    n = max(s["n"], 1)
    print(f"{field:18} {100*s['ok']/n:7.1f}% {100*s['missing']/n:7.1f}% {100*s['wrong']/n:7.1f}%")
````

````python
import gc
import json
import re
from copy import deepcopy
from pathlib import Path

import torch
from unsloth import FastLanguageModel

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)

# Header fields: trust v3, but if blank use v1
FALLBACK_FIELDS = ["vendor", "vendor_gstin", "tax", "irn"]

def parse_json(text):
    text = (text or "").strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None

def is_blank(v):
    if v is None:
        return True
    if isinstance(v, str) and not v.strip():
        return True
    return False

def add_pan_from_gstin(summary):
    summary = dict(summary or {})
    for src, dst in (("vendor_gstin", "vendor_pan"), ("customer_gstin", "customer_pan")):
        g = re.sub(r"[^A-Z0-9]", "", str(summary.get(src) or "").upper())
        if len(g) == 15:
            summary[dst] = g[2:12]
        else:
            summary.setdefault(dst, None)
    return summary

def stitch(v3_obj, v1_obj):
    if not isinstance(v3_obj, dict):
        v3_obj = {}
    if not isinstance(v1_obj, dict):
        v1_obj = {}
    out = deepcopy(v3_obj) if v3_obj else {"invoice_summary": {}, "line_items": v1_obj.get("line_items")}
    s3 = out.setdefault("invoice_summary", {})
    s1 = (v1_obj.get("invoice_summary") or {}) if isinstance(v1_obj, dict) else {}
    used = []
    for field in FALLBACK_FIELDS:
        if is_blank(s3.get(field)) and not is_blank(s1.get(field)):
            s3[field] = s1[field]
            used.append(field)
    s3 = add_pan_from_gstin(s3)
    out["invoice_summary"] = s3
    if not out.get("line_items") and v1_obj.get("line_items"):
        out["line_items"] = v1_obj["line_items"]
        used.append("line_items")
    return out, used

def generate(model, tokenizer, ocr_text):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr_text.strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=2048, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    return raw, parse_json(raw)

# Load base once, attach both adapters
for name in ("model", "tokenizer"):
    if name in globals():
        del globals()[name]
gc.collect()
torch.cuda.empty_cache()

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-1.5B-Instruct",
    max_seq_length=8192,
    dtype=None,
    load_in_4bit=True,
)
model.load_adapter("/content/drive/MyDrive/invoice_lora_v3", adapter_name="v3")
model.load_adapter("/content/drive/MyDrive/invoice_lora", adapter_name="v1")
FastLanguageModel.for_inference(model)

# One short held-out invoice
row = None
with Path("/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl").open(encoding="utf-8") as f:
    for line in f:
        cand = json.loads(line)
        if len(cand.get("ocr_text") or "") <= 6000:
            row = cand
            break
assert row is not None

model.set_adapter("v3")
raw_v3, obj_v3 = generate(model, tokenizer, row["ocr_text"])

model.set_adapter("v1")
raw_v1, obj_v1 = generate(model, tokenizer, row["ocr_text"])

hybrid, used = stitch(obj_v3, obj_v1)

print("blob", row["blob_path"])
print("fallback fields used from v1:", used or "(none)")
print("\n===== V3 =====\n")
print(json.dumps(obj_v3, ensure_ascii=False, indent=2) if obj_v3 else raw_v3)
print("\n===== V1 =====\n")
print(json.dumps(obj_v1, ensure_ascii=False, indent=2) if obj_v1 else raw_v1)
print("\n===== HYBRID =====\n")
print(json.dumps(hybrid, ensure_ascii=False, indent=2))
print("\n===== GEMINI =====\n")
print(json.dumps(row["gemini_json"], ensure_ascii=False, indent=2))
````

````python
import gc
import json
import re
from copy import deepcopy
from pathlib import Path

import torch
from unsloth import FastLanguageModel

# ---------- config ----------
V3_PATH = "/content/drive/MyDrive/invoice_lora_v3"
V1_PATH = "/content/drive/MyDrive/invoice_lora"
VAL_PATH = "/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl"
MAX_SEQ = 8192

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)

HEADER_FIELDS = ["vendor", "vendor_gstin", "tax", "irn"]
GSTIN_RE = re.compile(r"[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]")
IRN_RE = re.compile(r"\b[a-fA-F0-9]{64}\b")

# ---------- helpers ----------
def is_blank(v):
    if v is None:
        return True
    if isinstance(v, str):
        t = v.strip()
        if not t:
            return True
        if t.lower() in {"none", "null", "nan", "n/a", "-"}:
            return True
    return False

def norm_gstin(v):
    return re.sub(r"[^A-Z0-9]", "", str(v or "").upper())

def valid_gstin(v):
    g = norm_gstin(v)
    return bool(GSTIN_RE.fullmatch(g)), g

def parse_json(text):
    text = (text or "").strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None

def generate(model, tokenizer, ocr_text, max_new_tokens=2048):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr_text.strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=max_new_tokens, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    return raw, parse_json(raw)

def regex_fill(summary, ocr_text, used):
    summary = dict(summary or {})
    ocr = ocr_text or ""
    compact = re.sub(r"\s+", "", ocr.upper())

    # GSTINs in OCR order
    found = []
    for m in GSTIN_RE.finditer(re.sub(r"[^A-Z0-9]", "", ocr.upper())):
        g = m.group()
        if g not in found:
            found.append(g)

    ok, cur = valid_gstin(summary.get("vendor_gstin"))
    if (not ok) and found:
        summary["vendor_gstin"] = found[0]
        used.append("regex:vendor_gstin")

    ok_c, _ = valid_gstin(summary.get("customer_gstin"))
    if (not ok_c) and len(found) >= 2:
        # second distinct GSTIN often buyer; only fill if blank/invalid
        summary["customer_gstin"] = found[1]
        used.append("regex:customer_gstin")
    elif (not ok_c) and len(found) == 1 and found[0] != norm_gstin(summary.get("vendor_gstin")):
        summary["customer_gstin"] = found[0]
        used.append("regex:customer_gstin")

    if is_blank(summary.get("irn")):
        m = IRN_RE.search(ocr)
        if m:
            summary["irn"] = m.group().lower()
            used.append("regex:irn")

    return summary

def needs_header_fallback(summary):
    summary = summary or {}
    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(summary.get(k))
            if not ok:
                return True
        elif is_blank(summary.get(k)):
            return True
    return False

def merge_header_from_v1(base_obj, v1_obj, used):
    out = deepcopy(base_obj) if isinstance(base_obj, dict) else {"invoice_summary": {}, "line_items": None}
    s = out.setdefault("invoice_summary", {})
    s1 = ((v1_obj or {}).get("invoice_summary") or {}) if isinstance(v1_obj, dict) else {}

    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(s.get(k))
            ok1, g1 = valid_gstin(s1.get(k))
            if (not ok) and ok1:
                s[k] = g1
                used.append(f"v1:{k}")
        else:
            if is_blank(s.get(k)) and not is_blank(s1.get(k)):
                s[k] = s1[k]
                used.append(f"v1:{k}")

    out["invoice_summary"] = s
    # never take v1 line_items
    return out

def run_pipeline(model, tokenizer, ocr_text):
    used = []

    model.set_adapter("v3")
    raw_v3, obj_v3 = generate(model, tokenizer, ocr_text)
    if not isinstance(obj_v3, dict):
        obj_v3 = {"invoice_summary": {}, "line_items": None, "_raw": raw_v3}

    summary = dict(obj_v3.get("invoice_summary") or {})
    summary = regex_fill(summary, ocr_text, used)
    obj_v3["invoice_summary"] = summary

    obj_v1 = None
    raw_v1 = None
    if needs_header_fallback(summary):
        model.set_adapter("v1")
        raw_v1, obj_v1 = generate(model, tokenizer, ocr_text)
        obj_v3 = merge_header_from_v1(obj_v3, obj_v1, used)

    return {
        "final": obj_v3,
        "v3_raw": raw_v3,
        "v1_raw": raw_v1,
        "v1_obj": obj_v1,
        "fallback_used": used,
    }

# ---------- load once ----------
for name in ("model", "tokenizer"):
    if name in globals():
        del globals()[name]
gc.collect()
torch.cuda.empty_cache()

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-1.5B-Instruct",
    max_seq_length=MAX_SEQ,
    dtype=None,
    load_in_4bit=True,
)
model.load_adapter(V3_PATH, adapter_name="v3")
model.load_adapter(V1_PATH, adapter_name="v1")
FastLanguageModel.for_inference(model)
print("loaded v3 + v1 adapters")

# ---------- one short val invoice demo ----------
row = None
with Path(VAL_PATH).open(encoding="utf-8") as f:
    for line in f:
        cand = json.loads(line)
        if len(cand.get("ocr_text") or "") <= 6000:
            row = cand
            break
assert row is not None, "no short val row"

result = run_pipeline(model, tokenizer, row["ocr_text"])

print("blob", row["blob_path"])
print("fallback_used", result["fallback_used"] or "(none)")
print("\n===== FINAL (v3 + regex + v1 header) =====\n")
print(json.dumps(result["final"], ensure_ascii=False, indent=2))
print("\n===== GEMINI =====\n")
print(json.dumps(row["gemini_json"], ensure_ascii=False, indent=2))
````

````python
"""
Pre-production validation for the invoice cascade:
  v3 (full JSON) -> regex (GSTIN/IRN) -> v1 (header fill-if-blank)

Run in Colab with Drive mounted and a GPU.
Does not train. Does not write Azure. Does not touch SAP.
"""

from __future__ import annotations

import gc
import json
import random
import re
from collections import defaultdict
from copy import deepcopy
from pathlib import Path

import torch
from unsloth import FastLanguageModel

# ---------- paths (Colab Drive) ----------
V3_PATH = "/content/drive/MyDrive/invoice_lora_v3"
V1_PATH = "/content/drive/MyDrive/invoice_lora"
VAL_PATH = "/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl"
OUT_PATH = "/content/drive/MyDrive/pipeline_validation_report.json"

MAX_SEQ = 8192
# Skip invoices whose prompt is too long for this version
MAX_PROMPT_TOKENS = 6000
# How many invoices per held-out vendor (same style as earlier scores)
PER_VENDOR = 2
MAX_VENDORS = 12
SEED = 3407

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)
HEADER_FIELDS = ["vendor", "vendor_gstin", "tax", "irn"]
SCORE_FIELDS = [
    "invoice_no",
    "invoice_date",
    "vendor",
    "customer",
    "vendor_gstin",
    "customer_gstin",
    "tax",
    "total",
    "irn",
]
STRICT_FIELDS = ["invoice_no", "vendor_gstin", "customer_gstin", "tax", "total"]

GSTIN_RE = re.compile(r"[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]")
IRN_RE = re.compile(r"\b[a-fA-F0-9]{64}\b")
SUMMARY_REQUIRED = {
    "invoice_no",
    "invoice_date",
    "vendor",
    "customer",
    "vendor_gstin",
    "customer_gstin",
    "customer_address",
    "irn",
    "tax",
    "total",
}


def is_blank(v) -> bool:
    if v is None:
        return True
    if isinstance(v, str):
        t = v.strip()
        if not t:
            return True
        if t.lower() in {"none", "null", "nan", "n/a", "-"}:
            return True
    return False


def norm(v) -> str:
    if v is None:
        return ""
    return re.sub(r"\s+", " ", str(v).strip().upper())


def norm_gstin(v) -> str:
    return re.sub(r"[^A-Z0-9]", "", str(v or "").upper())


def valid_gstin(v) -> tuple[bool, str]:
    g = norm_gstin(v)
    return bool(re.fullmatch(GSTIN_RE.pattern, g)), g


def parse_json(text):
    text = (text or "").strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None


def line_count(obj):
    if not isinstance(obj, dict):
        return None
    li = obj.get("line_items")
    if isinstance(li, dict) and isinstance(li.get("items"), list):
        return len(li["items"])
    if isinstance(li, list):
        return len(li)
    return None


def schema_ok(obj) -> tuple[bool, list[str]]:
    problems = []
    if not isinstance(obj, dict):
        return False, ["not_a_dict"]
    if "invoice_summary" not in obj:
        problems.append("missing_invoice_summary")
    if "line_items" not in obj:
        problems.append("missing_line_items")
    summary = obj.get("invoice_summary")
    if not isinstance(summary, dict):
        problems.append("invoice_summary_not_dict")
    else:
        missing = SUMMARY_REQUIRED - set(summary.keys())
        if missing:
            problems.append("summary_missing_keys:" + ",".join(sorted(missing)))
    li = obj.get("line_items")
    if isinstance(li, dict):
        if not isinstance(li.get("items"), list):
            problems.append("line_items.items_not_list")
    elif not isinstance(li, list):
        problems.append("line_items_bad_type")
    return len(problems) == 0, problems


def generate(model, tokenizer, ocr_text, max_new_tokens=2048):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr_text.strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False,
        use_cache=True,
    )
    raw = tokenizer.decode(
        out[0][inputs["input_ids"].shape[-1] :],
        skip_special_tokens=True,
    )
    return raw, parse_json(raw)


def regex_fill(summary, ocr_text, used):
    summary = dict(summary or {})
    ocr = ocr_text or ""
    found = []
    for m in GSTIN_RE.finditer(re.sub(r"[^A-Z0-9]", "", ocr.upper())):
        g = m.group()
        if g not in found:
            found.append(g)

    ok, _ = valid_gstin(summary.get("vendor_gstin"))
    if (not ok) and found:
        summary["vendor_gstin"] = found[0]
        used.append("regex:vendor_gstin")

    ok_c, _ = valid_gstin(summary.get("customer_gstin"))
    if (not ok_c) and len(found) >= 2:
        summary["customer_gstin"] = found[1]
        used.append("regex:customer_gstin")

    if is_blank(summary.get("irn")):
        m = IRN_RE.search(ocr)
        if m:
            summary["irn"] = m.group().lower()
            used.append("regex:irn")

    return summary


def needs_header_fallback(summary) -> bool:
    summary = summary or {}
    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(summary.get(k))
            if not ok:
                return True
        elif is_blank(summary.get(k)):
            return True
    return False


def merge_header_from_v1(base_obj, v1_obj, used):
    out = (
        deepcopy(base_obj)
        if isinstance(base_obj, dict)
        else {"invoice_summary": {}, "line_items": None}
    )
    s = out.setdefault("invoice_summary", {})
    s1 = ((v1_obj or {}).get("invoice_summary") or {}) if isinstance(v1_obj, dict) else {}

    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(s.get(k))
            ok1, g1 = valid_gstin(s1.get(k))
            if (not ok) and ok1:
                s[k] = g1
                used.append(f"v1:{k}")
        else:
            if is_blank(s.get(k)) and not is_blank(s1.get(k)):
                s[k] = s1[k]
                used.append(f"v1:{k}")

    out["invoice_summary"] = s
    return out


def run_pipeline(model, tokenizer, ocr_text):
    used = []
    model.set_adapter("v3")
    raw_v3, obj_v3 = generate(model, tokenizer, ocr_text)
    if not isinstance(obj_v3, dict):
        obj_v3 = {"invoice_summary": {}, "line_items": None}

    summary = dict(obj_v3.get("invoice_summary") or {})
    summary = regex_fill(summary, ocr_text, used)
    obj_v3["invoice_summary"] = summary

    if needs_header_fallback(summary):
        model.set_adapter("v1")
        _, obj_v1 = generate(model, tokenizer, ocr_text)
        obj_v3 = merge_header_from_v1(obj_v3, obj_v1, used)

    return obj_v3, used, raw_v3


def build_sample(val_path: Path):
    by_vendor = defaultdict(list)
    with val_path.open(encoding="utf-8") as f:
        for line in f:
            row = json.loads(line)
            by_vendor[row.get("vendor") or "UNKNOWN"].append(row)

    random.seed(SEED)
    sample = []
    for _, rows in sorted(by_vendor.items(), key=lambda x: -len(x[1])):
        if len(sample) >= MAX_VENDORS * PER_VENDOR:
            break
        random.shuffle(rows)
        sample.extend(rows[:PER_VENDOR])
    return sample


def main():
    from google.colab import drive

    drive.mount("/content/drive", force_remount=False)

    for name in ("model", "tokenizer"):
        if name in globals():
            del globals()[name]
    gc.collect()
    torch.cuda.empty_cache()

    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name="unsloth/Qwen2.5-1.5B-Instruct",
        max_seq_length=MAX_SEQ,
        dtype=None,
        load_in_4bit=True,
    )
    model.load_adapter(V3_PATH, adapter_name="v3")
    model.load_adapter(V1_PATH, adapter_name="v1")
    FastLanguageModel.for_inference(model)
    print("loaded cascade: v3 + regex + v1")

    sample = build_sample(Path(VAL_PATH))
    print("candidate invoices", len(sample))

    stats = {
        k: {"ok": 0, "missing": 0, "wrong": 0, "n": 0}
        for k in SCORE_FIELDS + ["line_count"]
    }
    schema_pass = 0
    strict_ok = 0
    scored = 0
    skipped_long = 0
    parse_fail = 0
    fallback_counts = defaultdict(int)
    rows_out = []

    for i, row in enumerate(sample, 1):
        ocr = (row.get("ocr_text") or "").strip()
        prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr
        text = tokenizer.apply_chat_template(
            [{"role": "user", "content": prompt}],
            tokenize=False,
            add_generation_prompt=True,
        )
        ntok = len(tokenizer(text, add_special_tokens=False)["input_ids"])
        if ntok > MAX_PROMPT_TOKENS:
            skipped_long += 1
            continue

        pred, used, raw = run_pipeline(model, tokenizer, ocr)
        for u in used:
            fallback_counts[u] += 1

        ok_schema, problems = schema_ok(pred)
        if ok_schema:
            schema_pass += 1
        if pred.get("invoice_summary") == {} and not parse_json(raw):
            parse_fail += 1

        gold = row.get("gemini_json") or {}
        gs = gold.get("invoice_summary") or {}
        ps = pred.get("invoice_summary") or {}

        field_row = {}
        for field in SCORE_FIELDS:
            g, p = norm(gs.get(field)), norm(ps.get(field))
            stats[field]["n"] += 1
            if is_blank(ps.get(field)):
                stats[field]["missing"] += 1
                status = "missing"
            elif p == g:
                stats[field]["ok"] += 1
                status = "ok"
            else:
                stats[field]["wrong"] += 1
                status = "wrong"
            field_row[field] = status

        gc_, pc_ = line_count(gold), line_count(pred)
        stats["line_count"]["n"] += 1
        if pc_ is None:
            stats["line_count"]["missing"] += 1
            field_row["line_count"] = "missing"
        elif pc_ == gc_:
            stats["line_count"]["ok"] += 1
            field_row["line_count"] = "ok"
        else:
            stats["line_count"]["wrong"] += 1
            field_row["line_count"] = "wrong"

        if all(
            (not is_blank(ps.get(k))) and norm(ps.get(k)) == norm(gs.get(k))
            for k in STRICT_FIELDS
        ):
            strict_ok += 1

        scored += 1
        rows_out.append(
            {
                "blob_path": row.get("blob_path"),
                "vendor": row.get("vendor"),
                "prompt_tokens": ntok,
                "schema_ok": ok_schema,
                "schema_problems": problems,
                "fallback_used": used,
                "fields": field_row,
            }
        )
        if scored % 5 == 0:
            print(f"scored {scored} | skipped_long {skipped_long}")

    total_ok = sum(stats[k]["ok"] for k in stats)
    total_n = sum(stats[k]["n"] for k in stats)
    overall = 100 * total_ok / max(total_n, 1)
    strict_pct = 100 * strict_ok / max(scored, 1)
    schema_pct = 100 * schema_pass / max(scored, 1)

    print("\n========== PIPELINE VALIDATION ==========")
    print("pipeline: v3 -> regex -> v1 header")
    print("scored invoices", scored)
    print("skipped (too long for this version)", skipped_long)
    print(f"overall field accuracy {overall:.1f}%")
    print(
        f"strict (invoice_no + both GSTINs + tax + total) {strict_pct:.1f}%"
    )
    print(f"schema valid {schema_pct:.1f}%")
    print(f"json parse fail (empty fallback obj) {parse_fail}")
    print("\nfallback usage:")
    if fallback_counts:
        for k, v in sorted(fallback_counts.items(), key=lambda x: -x[1]):
            print(f"  {k}: {v}")
    else:
        print("  (none)")

    print(f"\n{'field':18} {'accuracy':>8} {'missing':>8} {'wrong':>8}")
    for field in sorted(stats, key=lambda k: stats[k]["ok"] / max(stats[k]["n"], 1)):
        s = stats[field]
        n = max(s["n"], 1)
        print(
            f"{field:18} {100 * s['ok'] / n:7.1f}% "
            f"{100 * s['missing'] / n:7.1f}% "
            f"{100 * s['wrong'] / n:7.1f}%"
        )

    report = {
        "pipeline": "v3 -> regex -> v1 header",
        "scored": scored,
        "skipped_long": skipped_long,
        "overall_field_accuracy_pct": round(overall, 2),
        "strict_pct": round(strict_pct, 2),
        "schema_valid_pct": round(schema_pct, 2),
        "parse_fail": parse_fail,
        "fallback_counts": dict(fallback_counts),
        "fields": {
            k: {
                "accuracy_pct": round(100 * v["ok"] / max(v["n"], 1), 2),
                "missing_pct": round(100 * v["missing"] / max(v["n"], 1), 2),
                "wrong_pct": round(100 * v["wrong"] / max(v["n"], 1), 2),
                "n": v["n"],
            }
            for k, v in stats.items()
        },
        "rows": rows_out,
        "scope_note": ( `
            "Invoices with prompt tokens > "
            f"{MAX_PROMPT_TOKENS} are skipped; this version targets normal-length "
            "PDF Plumber OCR only."
        ),
    }
    Path(OUT_PATH).write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
    print("\nsaved report", OUT_PATH)
    print("=========================================")


if __name__ == "__main__":
    main()
````

````python
"""
Pre-production validation for the invoice cascade:
  v3 (full JSON) -> regex (GSTIN/IRN) -> v1 (header fill-if-blank)

Run in Colab with Drive mounted and a GPU.
Does not train. Does not write Azure. Does not touch SAP.
"""

from __future__ import annotations

import gc
import json
import random
import re
from collections import defaultdict
from copy import deepcopy
from pathlib import Path

import torch
from unsloth import FastLanguageModel

# ---------- paths (Colab Drive) ----------
V3_PATH = "/content/drive/MyDrive/invoice_lora_v3"
V1_PATH = "/content/drive/MyDrive/invoice_lora"
VAL_PATH = "/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl"
OUT_PATH = "/content/drive/MyDrive/pipeline_validation_report.json"

MAX_SEQ = 8192
# Skip invoices whose prompt is too long for this version
MAX_PROMPT_TOKENS = 6000
# How many invoices per held-out vendor (same style as earlier scores)
PER_VENDOR = 2
MAX_VENDORS = 12
SEED = 3407

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)
HEADER_FIELDS = ["vendor", "vendor_gstin", "tax", "irn"]
SCORE_FIELDS = [
    "invoice_no",
    "invoice_date",
    "vendor",
    "customer",
    "vendor_gstin",
    "customer_gstin",
    "tax",
    "total",
    "irn",
]
STRICT_FIELDS = ["invoice_no", "vendor_gstin", "customer_gstin", "tax", "total"]

GSTIN_RE = re.compile(r"\b\d{2}[A-Z]{5}\d{4}[A-Z][A-Z0-9]Z[A-Z0-9]\b", re.I)
IRN_RE = re.compile(r"\b[a-fA-F0-9]{64}\b")
# Reject tiny/stub OCR (splitextract ocr_quality idea / BUG-002)
MIN_OCR_CHARS = 200
SUMMARY_REQUIRED = {
    "invoice_no",
    "invoice_date",
    "vendor",
    "customer",
    "vendor_gstin",
    "customer_gstin",
    "customer_address",
    "irn",
    "tax",
    "total",
}


def is_blank(v) -> bool:
    if v is None:
        return True
    if isinstance(v, str):
        t = v.strip()
        if not t:
            return True
        if t.lower() in {"none", "null", "nan", "n/a", "-"}:
            return True
    return False


def norm(v) -> str:
    if v is None:
        return ""
    return re.sub(r"\s+", " ", str(v).strip().upper())


def clean_gstin(gstin_str) -> str | None:
    """From splitextract: normalize OCR GSTIN and validate shape."""
    if not gstin_str:
        return None
    cleaned = re.sub(r"[\s\-]", "", str(gstin_str).upper().strip())
    cleaned = cleaned.replace("l", "1")
    pat = r"^\d{2}[A-Z]{5}\d{4}[A-Z][A-Z0-9]Z[A-Z0-9]$"
    if re.match(pat, cleaned):
        return cleaned
    fixed = list(cleaned)
    for pos in (0, 1, 7, 8, 9, 10):
        if pos < len(fixed) and fixed[pos] == "O":
            fixed[pos] = "0"
    fixed_s = "".join(fixed)
    if re.match(pat, fixed_s):
        return fixed_s
    return None


def valid_gstin(v) -> tuple[bool, str]:
    g = clean_gstin(v) or ""
    return bool(g), g


def gstins_from_ocr(ocr_text: str) -> list[str]:
    """Distinct GSTINs in OCR order (word-boundary, not compacted)."""
    found: list[str] = []
    for m in GSTIN_RE.finditer(ocr_text or ""):
        g = clean_gstin(m.group())
        if g and g not in found:
            found.append(g)
    return found


def _gstin_near_label(ocr: str, gstin: str, labels: tuple[str, ...]) -> bool:
    """True if a label appears within ~80 chars before this GSTIN."""
    up = (ocr or "").upper()
    g = gstin.upper()
    idx = up.find(g)
    if idx < 0:
        return False
    window = up[max(0, idx - 80) : idx]
    return any(lab in window for lab in labels)


def extract_net_amount_from_ocr(ocr_text: str) -> str | None:
    """From splitextract: NET AMOUNT / GRAND TOTAL / payable (largest sane)."""
    if not ocr_text:
        return None
    patterns = [
        r"NET\s*AMOUNT[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"GRAND\s*TOTAL[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"Invoice\s+Total[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"TOTAL\s+AMOUNT[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"(?:Amount\s+)?Payable[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"BILL\s+AMOUNT[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
    ]
    values: list[float] = []
    for pattern in patterns:
        for match in re.finditer(pattern, ocr_text, re.IGNORECASE):
            pre = ocr_text[match.start() : match.start(1)]
            if re.search(r"\bIRN\b", pre, re.IGNORECASE):
                continue
            try:
                value = float(match.group(1).replace(",", ""))
            except ValueError:
                continue
            if value > 100:
                values.append(value)
    if not values:
        return None
    unique = sorted(set(values), reverse=True)
    largest = unique[0]
    if len(unique) >= 2:
        second = unique[1]
        if largest > second * 50 and second > 100:
            ratio = largest / second
            if 90 <= ratio <= 110:
                largest = second
    return f"{largest:.2f}"


def extract_tax_from_ocr(ocr_text: str) -> str | None:
    """Fill blank tax from TOTAL GST / CGST+SGST / IGST labels."""
    if not ocr_text:
        return None
    patterns = [
        r"(?:TOTAL\s*GST|GST\s*AMOUNT|TAX\s*AMOUNT)[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
        r"\bIGST[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)",
    ]
    for pattern in patterns:
        m = re.search(pattern, ocr_text, re.IGNORECASE)
        if not m:
            continue
        try:
            value = float(m.group(1).replace(",", ""))
        except ValueError:
            continue
        if value >= 0:
            return f"{value:.2f}"

    cgst = re.search(r"\bCGST[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)", ocr_text, re.I)
    sgst = re.search(r"\bSGST[^0-9]{0,15}([0-9][0-9,]*(?:\.\d{1,2})?)", ocr_text, re.I)
    if cgst and sgst:
        try:
            total = float(cgst.group(1).replace(",", "")) + float(
                sgst.group(1).replace(",", "")
            )
            if total >= 0:
                return f"{total:.2f}"
        except ValueError:
            pass
    return None


def extract_invoice_no_from_ocr(ocr_text: str) -> str | None:
    """Light invoice-no pull; skip GSTIN-like and phone-like tokens."""
    if not ocr_text:
        return None
    patterns = [
        r"\bINV(?:OICE)?\.?\s*NO\.?\s*[:\-]?\s*([A-Z0-9][A-Z0-9/\-]{2,24})",
        r"\bINVOICE\s*(?:NO\.?|NUMBER|#)?\s*[:\-]?\s*([A-Z0-9][A-Z0-9/\-]{2,24})",
        r"\bBILL\s*NO\.?\s*[:\-]?\s*([A-Z0-9][A-Z0-9/\-]{2,24})",
    ]
    for pattern in patterns:
        for m in re.finditer(pattern, ocr_text, re.IGNORECASE):
            cand = m.group(1).strip().rstrip(".,;")
            if clean_gstin(cand):
                continue
            digits = re.sub(r"\D", "", cand)
            if len(digits) == 10 and digits[0] in "6789":
                continue
            if re.fullmatch(r"\d{6,8}", cand):  # likely HSN
                continue
            return cand
    return None


def parse_json(text):
    text = (text or "").strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None


def line_count(obj):
    if not isinstance(obj, dict):
        return None
    li = obj.get("line_items")
    if isinstance(li, dict) and isinstance(li.get("items"), list):
        return len(li["items"])
    if isinstance(li, list):
        return len(li)
    return None


def schema_ok(obj) -> tuple[bool, list[str]]:
    problems = []
    if not isinstance(obj, dict):
        return False, ["not_a_dict"]
    if "invoice_summary" not in obj:
        problems.append("missing_invoice_summary")
    if "line_items" not in obj:
        problems.append("missing_line_items")
    summary = obj.get("invoice_summary")
    if not isinstance(summary, dict):
        problems.append("invoice_summary_not_dict")
    else:
        missing = SUMMARY_REQUIRED - set(summary.keys())
        if missing:
            problems.append("summary_missing_keys:" + ",".join(sorted(missing)))
    li = obj.get("line_items")
    if isinstance(li, dict):
        if not isinstance(li.get("items"), list):
            problems.append("line_items.items_not_list")
    elif not isinstance(li, list):
        problems.append("line_items_bad_type")
    return len(problems) == 0, problems


def generate(model, tokenizer, ocr_text, max_new_tokens=2048):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr_text.strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False,
        use_cache=True,
    )
    raw = tokenizer.decode(
        out[0][inputs["input_ids"].shape[-1] :],
        skip_special_tokens=True,
    )
    return raw, parse_json(raw)


def regex_fill(summary, ocr_text, used):
    """
    Post-process with splitextract-style OCR regex.
    - clean / assign GSTINs (including customer==vendor poison fix)
    - fill blank IRN, total, tax, invoice_no from OCR labels
    Never invent a GSTIN that is not in OCR.
    """
    summary = dict(summary or {})
    ocr = ocr_text or ""
    found = gstins_from_ocr(ocr)

    vendor_labels = (
        "VENDOR GSTIN",
        "SUPPLIER GSTIN",
        "SELLER GSTIN",
        "FROM GSTIN",
        "GSTIN OF SUPPLIER",
        "GSTIN:",
        "GST NO",
        "GSTNO",
    )
    customer_labels = (
        "CUSTOMER GSTIN",
        "BUYER GSTIN",
        "BILL TO GSTIN",
        "SHIP TO GSTIN",
        "CONSIGNEE GSTIN",
        "GSTIN OF RECIPIENT",
        "GSTIN OF BUYER",
        "TO GSTIN",
    )

    ok_v, cur_v = valid_gstin(summary.get("vendor_gstin"))
    ok_c, cur_c = valid_gstin(summary.get("customer_gstin"))

    # Prefer label-anchored GSTIN when present in OCR
    labeled_vendor = next(
        (g for g in found if _gstin_near_label(ocr, g, vendor_labels)), None
    )
    labeled_customer = next(
        (g for g in found if _gstin_near_label(ocr, g, customer_labels) and g != labeled_vendor),
        None,
    )

    if labeled_vendor and (not ok_v or cur_v not in found):
        summary["vendor_gstin"] = labeled_vendor
        used.append("regex:vendor_gstin")
        ok_v, cur_v = True, labeled_vendor
    elif (not ok_v) and found:
        summary["vendor_gstin"] = found[0]
        used.append("regex:vendor_gstin")
        ok_v, cur_v = True, found[0]
    elif ok_v and cur_v not in found and found:
        # Model GSTIN not grounded in OCR -> replace
        summary["vendor_gstin"] = labeled_vendor or found[0]
        used.append("regex:vendor_gstin_ungrounded")
        ok_v, cur_v = True, summary["vendor_gstin"]

    if labeled_customer and (not ok_c or cur_c == cur_v or cur_c not in found):
        summary["customer_gstin"] = labeled_customer
        used.append("regex:customer_gstin")
        ok_c, cur_c = True, labeled_customer
    elif (not ok_c) and len(found) >= 2:
        other = next((g for g in found if g != cur_v), found[1])
        summary["customer_gstin"] = other
        used.append("regex:customer_gstin")
        ok_c, cur_c = True, other
    elif ok_c and cur_c == cur_v and len(found) >= 2:
        # Poisoned duplicate: both set to vendor; pick the other OCR GSTIN
        other = next((g for g in found if g != cur_v), None)
        if other:
            summary["customer_gstin"] = other
            used.append("regex:customer_gstin_dedupe")
            ok_c, cur_c = True, other
    elif ok_c and cur_c not in found and len(found) >= 2:
        other = next((g for g in found if g != cur_v), found[-1])
        summary["customer_gstin"] = other
        used.append("regex:customer_gstin_ungrounded")

    if is_blank(summary.get("irn")):
        m = IRN_RE.search(ocr)
        if m:
            summary["irn"] = m.group().lower()
            used.append("regex:irn")

    if is_blank(summary.get("total")):
        amt = extract_net_amount_from_ocr(ocr)
        if amt:
            summary["total"] = amt
            used.append("regex:total")

    if is_blank(summary.get("tax")):
        tax = extract_tax_from_ocr(ocr)
        if tax:
            summary["tax"] = tax
            used.append("regex:tax")

    if is_blank(summary.get("invoice_no")):
        inv = extract_invoice_no_from_ocr(ocr)
        if inv:
            summary["invoice_no"] = inv
            used.append("regex:invoice_no")

    return summary


def needs_header_fallback(summary) -> bool:
    summary = summary or {}
    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(summary.get(k))
            if not ok:
                return True
        elif is_blank(summary.get(k)):
            return True
    return False


def merge_header_from_v1(base_obj, v1_obj, used):
    out = (
        deepcopy(base_obj)
        if isinstance(base_obj, dict)
        else {"invoice_summary": {}, "line_items": None}
    )
    s = out.setdefault("invoice_summary", {})
    s1 = ((v1_obj or {}).get("invoice_summary") or {}) if isinstance(v1_obj, dict) else {}

    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(s.get(k))
            ok1, g1 = valid_gstin(s1.get(k))
            if (not ok) and ok1:
                s[k] = g1
                used.append(f"v1:{k}")
        else:
            if is_blank(s.get(k)) and not is_blank(s1.get(k)):
                s[k] = s1[k]
                used.append(f"v1:{k}")

    out["invoice_summary"] = s
    return out


def run_pipeline(model, tokenizer, ocr_text):
    used = []
    ocr = (ocr_text or "").strip()
    if len(ocr) < MIN_OCR_CHARS:
        used.append("error:EMPTY_OCR")
        return (
            {"invoice_summary": {}, "line_items": None, "error": "EMPTY_OCR"},
            used,
            "",
        )

    model.set_adapter("v3")
    raw_v3, obj_v3 = generate(model, tokenizer, ocr)
    if not isinstance(obj_v3, dict):
        obj_v3 = {"invoice_summary": {}, "line_items": None}

    summary = dict(obj_v3.get("invoice_summary") or {})
    summary = regex_fill(summary, ocr, used)
    obj_v3["invoice_summary"] = summary

    if needs_header_fallback(summary):
        model.set_adapter("v1")
        _, obj_v1 = generate(model, tokenizer, ocr)
        obj_v3 = merge_header_from_v1(obj_v3, obj_v1, used)
        # Second regex pass: fix GSTIN dedupe / totals after v1
        summary = regex_fill(dict(obj_v3.get("invoice_summary") or {}), ocr, used)
        obj_v3["invoice_summary"] = summary

    return obj_v3, used, raw_v3


def build_sample(val_path: Path):
    by_vendor = defaultdict(list)
    with val_path.open(encoding="utf-8") as f:
        for line in f:
            row = json.loads(line)
            by_vendor[row.get("vendor") or "UNKNOWN"].append(row)

    random.seed(SEED)
    sample = []
    for _, rows in sorted(by_vendor.items(), key=lambda x: -len(x[1])):
        if len(sample) >= MAX_VENDORS * PER_VENDOR:
            break
        random.shuffle(rows)
        sample.extend(rows[:PER_VENDOR])
    return sample


def main():
    from google.colab import drive

    drive.mount("/content/drive", force_remount=False)

    for name in ("model", "tokenizer"):
        if name in globals():
            del globals()[name]
    gc.collect()
    torch.cuda.empty_cache()

    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name="unsloth/Qwen2.5-1.5B-Instruct",
        max_seq_length=MAX_SEQ,
        dtype=None,
        load_in_4bit=True,
    )
    model.load_adapter(V3_PATH, adapter_name="v3")
    model.load_adapter(V1_PATH, adapter_name="v1")
    FastLanguageModel.for_inference(model)
    print("loaded cascade: v3 + regex + v1")

    sample = build_sample(Path(VAL_PATH))
    print("candidate invoices", len(sample))

    stats = {
        k: {"ok": 0, "missing": 0, "wrong": 0, "n": 0}
        for k in SCORE_FIELDS + ["line_count"]
    }
    schema_pass = 0
    strict_ok = 0
    scored = 0
    skipped_long = 0
    parse_fail = 0
    fallback_counts = defaultdict(int)
    rows_out = []

    for i, row in enumerate(sample, 1):
        ocr = (row.get("ocr_text") or "").strip()
        prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr
        text = tokenizer.apply_chat_template(
            [{"role": "user", "content": prompt}],
            tokenize=False,
            add_generation_prompt=True,
        )
        ntok = len(tokenizer(text, add_special_tokens=False)["input_ids"])
        if ntok > MAX_PROMPT_TOKENS:
            skipped_long += 1
            continue

        pred, used, raw = run_pipeline(model, tokenizer, ocr)
        for u in used:
            fallback_counts[u] += 1

        ok_schema, problems = schema_ok(pred)
        if ok_schema:
            schema_pass += 1
        if pred.get("invoice_summary") == {} and not parse_json(raw):
            parse_fail += 1

        gold = row.get("gemini_json") or {}
        gs = gold.get("invoice_summary") or {}
        ps = pred.get("invoice_summary") or {}

        field_row = {}
        for field in SCORE_FIELDS:
            g, p = norm(gs.get(field)), norm(ps.get(field))
            stats[field]["n"] += 1
            if is_blank(ps.get(field)):
                stats[field]["missing"] += 1
                status = "missing"
            elif p == g:
                stats[field]["ok"] += 1
                status = "ok"
            else:
                stats[field]["wrong"] += 1
                status = "wrong"
            field_row[field] = status

        gc_, pc_ = line_count(gold), line_count(pred)
        stats["line_count"]["n"] += 1
        if pc_ is None:
            stats["line_count"]["missing"] += 1
            field_row["line_count"] = "missing"
        elif pc_ == gc_:
            stats["line_count"]["ok"] += 1
            field_row["line_count"] = "ok"
        else:
            stats["line_count"]["wrong"] += 1
            field_row["line_count"] = "wrong"

        if all(
            (not is_blank(ps.get(k))) and norm(ps.get(k)) == norm(gs.get(k))
            for k in STRICT_FIELDS
        ):
            strict_ok += 1

        scored += 1
        rows_out.append(
            {
                "blob_path": row.get("blob_path"),
                "vendor": row.get("vendor"),
                "prompt_tokens": ntok,
                "schema_ok": ok_schema,
                "schema_problems": problems,
                "fallback_used": used,
                "fields": field_row,
            }
        )
        if scored % 5 == 0:
            print(f"scored {scored} | skipped_long {skipped_long}")

    total_ok = sum(stats[k]["ok"] for k in stats)
    total_n = sum(stats[k]["n"] for k in stats)
    overall = 100 * total_ok / max(total_n, 1)
    strict_pct = 100 * strict_ok / max(scored, 1)
    schema_pct = 100 * schema_pass / max(scored, 1)

    print("\n========== PIPELINE VALIDATION ==========")
    print("pipeline: v3 -> regex -> v1 header")
    print("scored invoices", scored)
    print("skipped (too long for this version)", skipped_long)
    print(f"overall field accuracy {overall:.1f}%")
    print(
        f"strict (invoice_no + both GSTINs + tax + total) {strict_pct:.1f}%"
    )
    print(f"schema valid {schema_pct:.1f}%")
    print(f"json parse fail (empty fallback obj) {parse_fail}")
    print("\nfallback usage:")
    if fallback_counts:
        for k, v in sorted(fallback_counts.items(), key=lambda x: -x[1]):
            print(f"  {k}: {v}")
    else:
        print("  (none)")

    print(f"\n{'field':18} {'accuracy':>8} {'missing':>8} {'wrong':>8}")
    for field in sorted(stats, key=lambda k: stats[k]["ok"] / max(stats[k]["n"], 1)):
        s = stats[field]
        n = max(s["n"], 1)
        print(
            f"{field:18} {100 * s['ok'] / n:7.1f}% "
            f"{100 * s['missing'] / n:7.1f}% "
            f"{100 * s['wrong'] / n:7.1f}%"
        )

    report = {
        "pipeline": "v3 -> regex -> v1 header",
        "scored": scored,
        "skipped_long": skipped_long,
        "overall_field_accuracy_pct": round(overall, 2),
        "strict_pct": round(strict_pct, 2),
        "schema_valid_pct": round(schema_pct, 2),
        "parse_fail": parse_fail,
        "fallback_counts": dict(fallback_counts),
        "fields": {
            k: {
                "accuracy_pct": round(100 * v["ok"] / max(v["n"], 1), 2),
                "missing_pct": round(100 * v["missing"] / max(v["n"], 1), 2),
                "wrong_pct": round(100 * v["wrong"] / max(v["n"], 1), 2),
                "n": v["n"],
            }
            for k, v in stats.items()
        },
        "rows": rows_out,
        "scope_note": (
            "Invoices with prompt tokens > "
            f"{MAX_PROMPT_TOKENS} are skipped; this version targets normal-length "
            "PDF Plumber OCR only."
        ),
    }
    Path(OUT_PATH).write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
    print("\nsaved report", OUT_PATH)
    print("=========================================")


if __name__ == "__main__":
    main()
````

````python
# IRN script (run AFTER v3 + regex + v1). Does not change regex / v1 / v3.
# Any 64-character hexadecimal string is treated as IRN.

import re

IRN_LENGTH = 64
HEX_CHARS = set("0123456789abcdefABCDEF")
IRN_CONTIGUOUS_RE = re.compile(r"[a-fA-F0-9]{64}")
IRN_GAPPED_RE = re.compile(r"[a-fA-F0-9](?:[\s\-]*[a-fA-F0-9]){63}")
IRN_LABEL_RE = re.compile(r"\bIRN(?:\s*(?:NO\.?|NUMBER|#))?\s*[:.,\-]?\s*", re.I)
_STOP_FIELD_RE = re.compile(
    r"\b(?:ACK(?:\.?\s*NO)?|GSTIN|GST\s*NO|DL\.?\s*NO|CIN|"
    r"INV(?:OICE)?\s*(?:NO|NUM)|ORDER\s*DATE|BILL\s*DT|"
    r"HSN|S\.?\s*N\.?)\b",
    re.I,
)


def _compact_hex(text):
    return "".join(ch.lower() for ch in text if ch in HEX_CHARS)


def _collect_hex_allowing_gaps(text):
    chars = []
    for ch in text:
        if ch in HEX_CHARS:
            chars.append(ch.lower())
            if len(chars) == IRN_LENGTH:
                break
        elif ch.isspace() or ch in "-:.,":
            continue
        else:
            break
    return "".join(chars)


def _line_is_hex_continuation(line):
    stripped = line.strip()
    if not stripped or stripped[0] not in HEX_CHARS:
        return False
    extra = sum(1 for ch in stripped if ch.isalpha() and ch.lower() not in "abcdef")
    return extra < 3


def find_irn_near_label(ocr_text):
    text = ocr_text or ""
    for match in IRN_LABEL_RE.finditer(text):
        rest = text[match.end():]
        stop = _STOP_FIELD_RE.search(rest)
        region = rest[:stop.start()] if stop else rest[:320]
        hexed = _collect_hex_allowing_gaps(region)
        if len(hexed) == IRN_LENGTH:
            return hexed
        lines = rest.splitlines()
        assembled = _collect_hex_allowing_gaps(lines[0] if lines else "")
        for line in lines[1:5]:
            if not line.strip():
                continue
            if not _line_is_hex_continuation(line):
                break
            assembled += _compact_hex(line)
            if len(assembled) >= IRN_LENGTH:
                return assembled[:IRN_LENGTH]
        if len(assembled) == IRN_LENGTH:
            return assembled
    return None


def extract_irn_64(ocr_text):
    text = ocr_text or ""
    found, seen = [], set()

    def add(v):
        if v and v not in seen:
            seen.add(v)
            found.append(v)

    add(find_irn_near_label(text))
    for m in IRN_CONTIGUOUS_RE.finditer(text):
        add(m.group().lower())
    for m in IRN_GAPPED_RE.finditer(text):
        add(_compact_hex(m.group()))
    return found[0] if found else None


def apply_irn_from_script(summary, ocr_text, used=None):
    out = dict(summary or {})
    used = used if used is not None else []
    found = extract_irn_64(ocr_text)
    if not found:
        return out
    current = out.get("irn")
    current_text = current.strip().lower() if isinstance(current, str) else ""
    if current_text != found:
        used.append("script:irn")
    out["irn"] = found
    return out


# Use on pipeline output (v3 -> regex -> v1 stays the same; IRN comes from this script at the end):
#   result = run_pipeline(model, tokenizer, ocr_text)
#   s = result["final"].setdefault("invoice_summary", {})
#   result["final"]["invoice_summary"] = apply_irn_from_script(s, ocr_text, result["fallback_used"])
#   print(result["final"]["invoice_summary"].get("irn"))

ocr_demo = "IRN:670a20b56d4546b638e56007520543bb5746a73542f614cf03b52cbeb417034cAck.No.:1326"
print(extract_irn_64(ocr_demo))
````

````python
# IRN accuracy: regex vs 64-hex script (final pipeline overlay)
# Run after Drive is mounted. Does not change v3 / regex / v1.

import json
import re
from collections import Counter
from pathlib import Path

IRN_LENGTH = 64
HEX_CHARS = set("0123456789abcdefABCDEF")
IRN_RE = re.compile(r"\b[a-fA-F0-9]{64}\b")  # existing regex (unchanged)
IRN_CONTIGUOUS_RE = re.compile(r"[a-fA-F0-9]{64}")
IRN_GAPPED_RE = re.compile(r"[a-fA-F0-9](?:[\s\-]*[a-fA-F0-9]){63}")
IRN_LABEL_RE = re.compile(r"\bIRN(?:\s*(?:NO\.?|NUMBER|#))?\s*[:.,\-]?\s*", re.I)
_STOP_FIELD_RE = re.compile(
    r"\b(?:ACK(?:\.?\s*NO)?|GSTIN|GST\s*NO|DL\.?\s*NO|CIN|"
    r"INV(?:OICE)?\s*(?:NO|NUM)|ORDER\s*DATE|BILL\s*DT|HSN|S\.?\s*N\.?)\b",
    re.I,
)
BLANK = {"", "none", "null", "nan", "n/a", "-", "na", "nil"}


def _compact_hex(text):
    return "".join(ch.lower() for ch in text if ch in HEX_CHARS)


def _collect_hex_allowing_gaps(text):
    chars = []
    for ch in text:
        if ch in HEX_CHARS:
            chars.append(ch.lower())
            if len(chars) == IRN_LENGTH:
                break
        elif ch.isspace() or ch in "-:.,":
            continue
        else:
            break
    return "".join(chars)


def extract_irn_64(ocr_text):
    text = ocr_text or ""
    found, seen = [], set()

    def add(v):
        if v and v not in seen:
            seen.add(v)
            found.append(v)

    for match in IRN_LABEL_RE.finditer(text):
        rest = text[match.end():]
        stop = _STOP_FIELD_RE.search(rest)
        region = rest[: stop.start()] if stop else rest[:320]
        hexed = _collect_hex_allowing_gaps(region)
        if len(hexed) == IRN_LENGTH:
            add(hexed)
            break
        lines = rest.splitlines()
        assembled = _collect_hex_allowing_gaps(lines[0] if lines else "")
        for line in lines[1:5]:
            if not line.strip():
                continue
            stripped = line.strip()
            extra = sum(1 for ch in stripped if ch.isalpha() and ch.lower() not in "abcdef")
            if extra >= 3 or not stripped or stripped[0] not in HEX_CHARS:
                break
            assembled += _compact_hex(line)
            if len(assembled) >= IRN_LENGTH:
                add(assembled[:IRN_LENGTH])
                break
        if len(assembled) == IRN_LENGTH:
            add(assembled)
            break
    for m in IRN_CONTIGUOUS_RE.finditer(text):
        add(m.group().lower())
    for m in IRN_GAPPED_RE.finditer(text):
        add(_compact_hex(m.group()))
    return found[0] if found else None


def gold_irn(row):
    summary = ((row.get("gemini_json") or {}).get("invoice_summary") or {})
    value = summary.get("irn")
    if value is None:
        return None
    text = str(value).strip()
    if not text or text.lower() in BLANK:
        return None
    compact = _compact_hex(text)
    return compact if len(compact) == 64 else text.lower()


def regex_irn(ocr):
    m = IRN_RE.search(ocr or "")
    return m.group().lower() if m else None


def gold_in_ocr(ocr, gold):
    if not gold or len(gold) != 64:
        return False
    return gold in re.sub(r"[^a-fA-F0-9]", "", ocr or "").lower()


def score(rows, pred_fn):
    c = Counter()
    gold_n = match_gold = 0
    for row in rows:
        ocr = row.get("ocr_text") or ""
        g = gold_irn(row)
        p = pred_fn(ocr)
        c["n"] += 1
        if g is None and p is None:
            c["both_blank"] += 1
        elif g is None:
            c["false_positive"] += 1
        elif p is None:
            c["missing"] += 1
        elif p == g:
            c["match"] += 1
        else:
            c["wrong"] += 1
        if g is not None:
            gold_n += 1
            if p == g:
                match_gold += 1
            if gold_in_ocr(ocr, g):
                c["gold_in_ocr"] += 1
                if p == g:
                    c["match_in_ocr"] += 1
    n = c["n"] or 1
    return {
        "invoices": c["n"],
        "gold_irn": gold_n,
        "match": c["match"],
        "missing": c["missing"],
        "wrong": c["wrong"],
        "both_blank": c["both_blank"],
        "false_positive": c["false_positive"],
        "acc_when_gold": round(100 * match_gold / gold_n, 2) if gold_n else None,
        "acc_all": round(100 * (c["match"] + c["both_blank"]) / n, 2),
        "acc_when_in_ocr": round(100 * c["match_in_ocr"] / c["gold_in_ocr"], 2) if c["gold_in_ocr"] else None,
    }


def load_jsonl(path):
    rows = []
    with path.open(encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


candidates = [
    Path("/content/drive/MyDrive/training_export"),
    Path("/content/drive/MyDrive/training_export/training_export"),
    Path(r"C:\Users\Prerana Bhalerao\OneDrive\Desktop\model train\training_export (2)\training_export"),
]
data_root = next((p for p in candidates if p.exists()), None)
assert data_root is not None, "Mount Drive or set training_export path"
print("data_root:", data_root)

val_path = data_root / "pdfplumber_val.jsonl"
train_files = sorted(p for p in data_root.glob("pdfplumber_train_batch_*.jsonl"))
splits = {
    "val": load_jsonl(val_path) if val_path.exists() else [],
    "train": [],
}
for p in train_files:
    print("loading", p.name)
    splits["train"].extend(load_jsonl(p))
splits["all"] = splits["val"] + splits["train"]

methods = [("regex_only", regex_irn), ("pipeline_irn_script", extract_irn_64)]
print("\n{:<8} {:>8} {:>8} {:>12} {:>12} {:>12} {:>14}".format(
    "split", "n", "gold", "regex_acc", "script_acc", "script_match", "acc_if_in_ocr"
))
print("-" * 86)
for name, rows in splits.items():
    stats = {m: score(rows, fn) for m, fn in methods}
    r, s = stats["regex_only"], stats["pipeline_irn_script"]
    print("{:<8} {:>8} {:>8} {:>11}% {:>11}% {:>12} {:>13}%".format(
        name, s["invoices"], s["gold_irn"], r["acc_when_gold"], s["acc_when_gold"],
        f"{s['match']}/{s['gold_irn']}", s["acc_when_in_ocr"],
    ))

print("\nFull pipeline IRN = 64-char hex script at the end (v3/regex/v1 unchanged).")
s_all = score(splits["all"], extract_irn_64)
print(
    f"TOTAL accuracy when gold IRN exists: {s_all['acc_when_gold']}%  "
    f"({s_all['match']}/{s_all['gold_irn']})"
)
print(f"TOTAL accuracy including blanks:     {s_all['acc_all']}%")
print(f"If gold IRN is present in OCR:       {s_all['acc_when_in_ocr']}%")
print("errors:", {k: s_all[k] for k in ("missing", "wrong", "false_positive", "both_blank")})
````

````python
# GSTIN validation + assignment script (run AFTER v3 + regex + v1)
# Does NOT change regex / v1 / v3. Overwrites vendor_gstin / customer_gstin at the end.
# Rules:
#  - find all valid 15-char GSTINs in OCR
#  - prefer Supplier/Seller/Vendor labels for vendor
#  - prefer Buyer/Bill To/Ship To/Customer labels for customer
#  - never allow vendor_gstin == customer_gstin if OCR has >= 2 GSTINs
#  - if ambiguous -> blank + needs_review

import re
from typing import Any

GSTIN_RE = re.compile(r"[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]")
GSTIN_FULL = re.compile(r"^[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]$")

VENDOR_LABEL_RE = re.compile(
    r"\b(?:"
    r"SUPPLIER(?:\s*GSTIN|\s*GST)?|"
    r"SELLER(?:\s*GSTIN|\s*GST)?|"
    r"VENDOR(?:\s*GSTIN|\s*GST)?|"
    r"FROM\s*GSTIN|"
    r"GSTIN\s*(?:OF\s*)?(?:SUPPLIER|SELLER|VENDOR)|"
    r"GST\s*(?:NO\.?|IN|NUMBER)?\s*(?:OF\s*)?(?:SUPPLIER|SELLER|VENDOR)|"
    r"GSTIN|GST\s*NO\.?|GST\s*NUMBER|GST\s*IN"
    r")\b",
    re.I,
)
CUSTOMER_LABEL_RE = re.compile(
    r"\b(?:"
    r"BUYER(?:\s*GSTIN|\s*GST)?|"
    r"CUSTOMER(?:\s*GSTIN|\s*GST)?|"
    r"BILL\s*TO|SHIP\s*TO|CONSIGNEE|TO\s*GSTIN|"
    r"GSTIN\s*(?:OF\s*)?(?:BUYER|CUSTOMER|BILL\s*TO|SHIP\s*TO)|"
    r"GST\s*(?:NO\.?|IN|NUMBER)?\s*(?:OF\s*)?(?:BUYER|CUSTOMER)"
    r")\b",
    re.I,
)
BLANK = {"", "none", "null", "nan", "n/a", "-", "na", "nil"}


def is_blank(v):
    return v is None or (isinstance(v, str) and v.strip().lower() in BLANK)


def norm_gstin(v):
    return re.sub(r"[^A-Z0-9]", "", str(v or "").upper())


def valid_gstin(v):
    g = norm_gstin(v)
    return bool(GSTIN_FULL.fullmatch(g)), g


def find_gstins_with_spans(ocr_text):
    text = ocr_text or ""
    compact_chars, compact_to_orig = [], []
    for i, ch in enumerate(text):
        if ch.isalnum():
            compact_chars.append(ch.upper())
            compact_to_orig.append(i)
    compact = "".join(compact_chars)
    found, seen = [], set()
    for m in GSTIN_RE.finditer(compact):
        g = m.group()
        if g in seen:
            continue
        seen.add(g)
        found.append((g, compact_to_orig[m.start()], compact_to_orig[m.end() - 1] + 1))
    return found


def _nearest_label_distance(pos, labels):
    best = None
    for m in labels:
        d = pos - m.end()
        if d < -40:
            continue
        score = abs(d) if d >= 0 else abs(d) + 20
        if best is None or score < best:
            best = score
    return best


def assign_gstins_from_ocr(ocr_text):
    text = ocr_text or ""
    gstins = find_gstins_with_spans(text)
    all_g = [g for g, _, _ in gstins]
    used = []
    vendor_labels = list(VENDOR_LABEL_RE.finditer(text))
    customer_labels = list(CUSTOMER_LABEL_RE.finditer(text))
    strong_customer = [m for m in customer_labels if re.search(r"BUYER|CUSTOMER|BILL\s*TO|SHIP\s*TO|CONSIGNEE", m.group(), re.I)]
    strong_vendor = [m for m in vendor_labels if re.search(r"SUPPLIER|SELLER|VENDOR|FROM\s*GSTIN", m.group(), re.I)]

    out = {
        "vendor_gstin": None,
        "customer_gstin": None,
        "needs_review": False,
        "reason": None,
        "all_gstins": all_g,
        "used": used,
    }
    if not gstins:
        out["needs_review"] = True
        out["reason"] = "no_gstin_in_ocr"
        used.append("gstin_script:none")
        return out

    if len(gstins) == 1:
        g, pos, _ = gstins[0]
        out["vendor_gstin"] = g
        out["needs_review"] = True
        out["reason"] = "only_one_gstin"
        used.append("gstin_script:single_vendor")
        return out

    def scores(pos):
        vd = _nearest_label_distance(pos, strong_vendor or vendor_labels)
        cd = _nearest_label_distance(pos, strong_customer or customer_labels)
        return (max(0.0, 1000.0 - vd) if vd is not None else 0.0,
                max(0.0, 1000.0 - cd) if cd is not None else 0.0)

    scored = []
    for g, pos, _ in gstins:
        v_s, c_s = scores(pos)
        idx = all_g.index(g)
        v_s += max(0.0, 5.0 - idx)
        c_s += idx * 1.5
        scored.append({"gstin": g, "pos": pos, "v": v_s, "c": c_s})

    by_v = sorted(scored, key=lambda x: (-x["v"], x["pos"]))
    vendor = by_v[0]["gstin"]
    remaining = [x for x in scored if x["gstin"] != vendor]
    by_c = sorted(remaining, key=lambda x: (-x["c"], -x["pos"]))
    customer = by_c[0]["gstin"]

    if by_v[0]["v"] < 50 and by_c[0]["c"] < 50:
        vendor, customer = all_g[0], all_g[1]
        used.append("gstin_script:order_fallback")
        out["needs_review"] = True
        out["reason"] = "ambiguous_label_scores"
    else:
        used.append("gstin_script:label_assign")

    if vendor == customer:
        other = next((g for g in all_g if g != vendor), None)
        if other:
            customer = other
            used.append("gstin_script:force_split")
            out["needs_review"] = True
            out["reason"] = "forced_split_equal"

    out["vendor_gstin"] = vendor
    out["customer_gstin"] = customer
    if out["vendor_gstin"] == out["customer_gstin"] and len(all_g) >= 2:
        out["customer_gstin"] = next(g for g in all_g if g != out["vendor_gstin"])
        used.append("gstin_script:final_split")
        out["needs_review"] = True
    if out["vendor_gstin"] is None or out["customer_gstin"] is None:
        out["needs_review"] = True
        out["reason"] = out["reason"] or "missing_party_gstin"
    return out


def apply_gstin_from_script(summary, ocr_text, used=None):
    out = dict(summary or {})
    used = used if used is not None else []
    a = assign_gstins_from_ocr(ocr_text)
    used.extend(a["used"])
    if a["vendor_gstin"]:
        if norm_gstin(out.get("vendor_gstin")) != a["vendor_gstin"]:
            used.append("script:vendor_gstin")
        out["vendor_gstin"] = a["vendor_gstin"]
    if a["customer_gstin"]:
        if norm_gstin(out.get("customer_gstin")) != a["customer_gstin"]:
            used.append("script:customer_gstin")
        out["customer_gstin"] = a["customer_gstin"]
    ok_v, vg = valid_gstin(out.get("vendor_gstin"))
    ok_c, cg = valid_gstin(out.get("customer_gstin"))
    if ok_v and ok_c and vg == cg and len(a["all_gstins"]) >= 2:
        out["customer_gstin"] = next(g for g in a["all_gstins"] if g != vg)
        used.append("script:force_unequal")
        a["needs_review"] = True
        a["reason"] = a["reason"] or "forced_unequal"
    return out, a["needs_review"], a["reason"]


# ---- demo ----
ocr_demo = """
HINDUSTAN PHARMA DISTRIBUTORS PVT LTD.
Supplier GSTIN : 29AACCD7912K1ZD
Bill To: ASTER HOSPITAL
Customer GSTIN : 29AAFCH9013A1ZV
Inv No: B159643
"""
demo = assign_gstins_from_ocr(ocr_demo)
print(demo)

# Use after pipeline:
#   result = run_pipeline(model, tokenizer, ocr_text)
#   s = result["final"]["invoice_summary"] if "final" in result else result["invoice_summary"]
#   s, needs_review, reason = apply_gstin_from_script(s, ocr_text, result.get("fallback_used", []))
#   print(s["vendor_gstin"], s["customer_gstin"], needs_review, reason)
````

````python
# ============================================================
# Train tiny GSTIN repair LoRA (vendor vs customer ONLY)
# Reuses same base: unsloth/Qwen2.5-1.5B-Instruct
# Saves to: /content/drive/MyDrive/invoice_adapters_all/invoice_lora_gstin_repair
# ============================================================

from pathlib import Path
from google.colab import drive
import json, re, random
import gc as py_gc
import torch
from datasets import Dataset
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only
from trl import SFTTrainer, SFTConfig

drive.mount("/content/drive", force_remount=False)

DATA_ROOT = Path("/content/drive/MyDrive/training_export/training_export")
OUT_DIR = Path("/content/drive/MyDrive/invoice_adapters_all/invoice_lora_gstin_repair")
OUT_DIR.mkdir(parents=True, exist_ok=True)

GSTIN_RE = re.compile(r"[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]")
GSTIN_FULL = re.compile(r"^[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]$")
BLANK = {"", "none", "null", "nan", "n/a", "-", "na", "nil"}
MAX_SEQ = 4096
SEED = 3407
MAX_TRAIN_ROWS = 8000  # raise if you want more


def norm_gstin(v):
    g = re.sub(r"[^A-Z0-9]", "", str(v or "").upper())
    return g if GSTIN_FULL.fullmatch(g) else ""


def find_gstins(ocr):
    compact = re.sub(r"[^A-Z0-9]", "", (ocr or "").upper())
    found, seen = [], set()
    for m in GSTIN_RE.finditer(compact):
        g = m.group()
        if g not in seen:
            seen.add(g)
            found.append(g)
    return found


def load_jsonl(path):
    rows = []
    with path.open(encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


# ---- build clean GSTIN-only train set ----
train_files = sorted(DATA_ROOT.glob("pdfplumber_train_batch_*.jsonl"))
assert train_files, f"no train jsonl in {DATA_ROOT}"

examples = []
for path in train_files:
    for row in load_jsonl(path):
        ocr = (row.get("ocr_text") or "").strip()
        if len(ocr) < 200 or len(ocr) > 12000:
            continue
        gold = ((row.get("gemini_json") or {}).get("invoice_summary") or {})
        # NOTE: do NOT name this `gc` — that shadows the gc module
        gv, gcust = norm_gstin(gold.get("vendor_gstin")), norm_gstin(gold.get("customer_gstin"))
        found = find_gstins(ocr)
        # keep only clean, grounded, distinct pairs
        if not (gv and gcust and gv != gcust):
            continue
        if gv not in found or gcust not in found:
            continue
        if len(found) < 2:
            continue
        # shuffle candidate order so model learns role, not list order
        cands = found[:]
        random.Random(hash(gv + gcust) % (10**9)).shuffle(cands)
        user = (
            "Assign which GSTIN is vendor and which is customer from this invoice OCR.\n"
            "Use ONLY GSTINs from the candidate list. Reply with JSON only.\n\n"
            f"Candidates: {json.dumps(cands)}\n\n"
            f"OCR text:\n{ocr[:8000]}"
        )
        assistant = json.dumps(
            {"vendor_gstin": gv, "customer_gstin": gcust},
            ensure_ascii=False,
        )
        examples.append({
            "conversations": [
                {"role": "user", "content": user},
                {"role": "assistant", "content": assistant},
            ]
        })
        if len(examples) >= MAX_TRAIN_ROWS:
            break
    if len(examples) >= MAX_TRAIN_ROWS:
        break

random.Random(SEED).shuffle(examples)
print(f"clean GSTIN train examples: {len(examples)}")
assert len(examples) >= 200, "too few clean rows — check gold / paths"

# save dataset for reuse
ds_path = OUT_DIR / "gstin_train_examples.jsonl"
with ds_path.open("w", encoding="utf-8") as f:
    for ex in examples:
        f.write(json.dumps(ex, ensure_ascii=False) + "\n")
print("saved", ds_path)

# ---- load same base + new LoRA ----
for name in ("model", "tokenizer", "trainer"):
    if name in globals():
        del globals()[name]
py_gc.collect()
torch.cuda.empty_cache()

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-1.5B-Instruct",
    max_seq_length=MAX_SEQ,
    dtype=None,
    load_in_4bit=True,
)
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=SEED,
)

def to_text(batch):
    texts = [
        tokenizer.apply_chat_template(convo, tokenize=False, add_generation_prompt=False)
        for convo in batch["conversations"]
    ]
    return {"text": texts}

dataset = Dataset.from_list(examples).map(to_text, batched=True, remove_columns=["conversations"])

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    args=SFTConfig(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        warmup_steps=20,
        num_train_epochs=1,
        learning_rate=2e-4,
        fp16=not is_bfloat16_supported(),
        bf16=is_bfloat16_supported(),
        logging_steps=10,
        optim="adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="linear",
        seed=SEED,
        output_dir=str(OUT_DIR / "checkpoints"),
        report_to="none",
        max_seq_length=MAX_SEQ,
        dataset_text_field="text",
    ),
)
trainer = train_on_responses_only(
    trainer,
    instruction_part="<|im_start|>user\n",
    response_part="<|im_start|>assistant\n",
)

print("starting GSTIN repair training...")
trainer.train()
model.save_pretrained(str(OUT_DIR))
tokenizer.save_pretrained(str(OUT_DIR))
print("saved adapter to", OUT_DIR)
````

````python
# ============================================================
# Use gstin_repair AFTER GSTIN script (same loaded base)
# Requires: model/tokenizer already loaded with v3+v1, and assign_gstins_from_ocr
# ============================================================

from pathlib import Path
import json
import re

GSTIN_ADAPTER = "/content/drive/MyDrive/invoice_adapters_all/invoice_lora_gstin_repair"
assert Path(GSTIN_ADAPTER, "adapter_config.json").exists(), GSTIN_ADAPTER

# load once on the same model object
try:
    model.load_adapter(GSTIN_ADAPTER, adapter_name="gstin_repair")
except Exception:
    # already loaded
    pass

def run_gstin_repair(model, tokenizer, ocr_text, candidates):
    prompt = (
        "Assign which GSTIN is vendor and which is customer from this invoice OCR.\n"
        "Use ONLY GSTINs from the candidate list. Reply with JSON only.\n\n"
        f"Candidates: {json.dumps(candidates)}\n\n"
        f"OCR text:\n{(ocr_text or '')[:8000]}"
    )
    model.set_adapter("gstin_repair")
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False,
        add_generation_prompt=True,
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=128, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    raw = raw.strip()
    if raw.startswith("```"):
        raw = re.sub(r"^```(?:json)?\s*", "", raw)
        raw = re.sub(r"\s*```$", "", raw)
    try:
        obj = json.loads(raw)
    except Exception:
        return None, raw
    return obj, raw


def apply_gstin_with_repair(summary, ocr_text, used=None):
    """Script finds GSTINs; tiny adapter assigns vendor vs customer when >=2."""
    used = used if used is not None else []
    out = dict(summary or {})
    a = assign_gstins_from_ocr(ocr_text)
    used.extend(a.get("used") or [])
    cands = a.get("all_gstins") or []

    if len(cands) < 2:
        # keep script result (single / none)
        if a.get("vendor_gstin"):
            out["vendor_gstin"] = a["vendor_gstin"]
        if a.get("customer_gstin"):
            out["customer_gstin"] = a["customer_gstin"]
        return out, a.get("needs_review", True), a.get("reason"), used

    obj, raw = run_gstin_repair(model, tokenizer, ocr_text, cands)
    if not isinstance(obj, dict):
        # fallback to script assignment
        out["vendor_gstin"] = a.get("vendor_gstin")
        out["customer_gstin"] = a.get("customer_gstin")
        used.append("gstin_repair:parse_fail")
        return out, True, "gstin_repair_parse_fail", used

    vg = norm_gstin(obj.get("vendor_gstin"))
    cg = norm_gstin(obj.get("customer_gstin"))
    # ground to candidates only
    if vg not in cands or cg not in cands or vg == cg:
        out["vendor_gstin"] = a.get("vendor_gstin")
        out["customer_gstin"] = a.get("customer_gstin")
        used.append("gstin_repair:ungrounded")
        return out, True, "gstin_repair_ungrounded", used

    out["vendor_gstin"] = vg
    out["customer_gstin"] = cg
    used.append("gstin_repair:assign")
    return out, False, None, used


# demo on one val row
val = Path("/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl")
row = json.loads(next(val.open(encoding="utf-8")))
ocr = row["ocr_text"]
summary = ((row.get("gemini_json") or {}).get("invoice_summary") or {})
fixed, review, reason, used = apply_gstin_with_repair({}, ocr, [])
print("gold vendor/customer:", norm_gstin(summary.get("vendor_gstin")), norm_gstin(summary.get("customer_gstin")))
print("pred vendor/customer:", fixed.get("vendor_gstin"), fixed.get("customer_gstin"))
print("review/reason/used:", review, reason, used)
````

````python
# ============================================================
# QUICK ACCURACY — 20 invoices only
# Pipeline: v3 -> regex -> v1 -> IRN script -> GSTIN script -> gstin_repair
# ============================================================

from pathlib import Path
from google.colab import drive
import json, re, gc as py_gc
from copy import deepcopy
import torch
from unsloth import FastLanguageModel
from peft import PeftModel
from safetensors import safe_open

drive.mount("/content/drive", force_remount=False)

LIMIT = 20
MAX_PROMPT_TOKENS = 6000
MAX_SEQ = 8192

V3_PATH = "/content/drive/MyDrive/invoice_adapters_all/invoice_lora_v3"
V1_PATH = "/content/drive/MyDrive/invoice_adapters_all/invoice_lora"
GSTIN_PATH = "/content/drive/MyDrive/invoice_adapters_all/invoice_lora_gstin_repair"

VAL_CANDIDATES = [
    Path("/content/drive/MyDrive/training_export/training_export/pdfplumber_val.jsonl"),
    Path("/content/drive/MyDrive/training_export/pdfplumber_val.jsonl"),
]
val_path = next((p for p in VAL_CANDIDATES if p.exists()), None)
if val_path is None:
    hits = list(Path("/content/drive/MyDrive").rglob("pdfplumber_val.jsonl"))
    assert hits, "val jsonl not found"
    val_path = hits[0]
print("VAL_PATH =", val_path)

SCORE_FIELDS = [
    "invoice_no", "invoice_date", "vendor", "customer",
    "vendor_gstin", "customer_gstin", "tax", "total", "irn",
]
STRICT_FIELDS = ["invoice_no", "vendor_gstin", "customer_gstin", "tax", "total"]
HEADER_FIELDS = ["vendor", "vendor_gstin", "tax", "irn"]
BLANK = {"", "none", "null", "nan", "n/a", "-", "na", "nil"}
HEX = set("0123456789abcdefABCDEF")
GSTIN_RE = re.compile(r"[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]")
GSTIN_FULL = re.compile(r"^[0-9]{2}[A-Z]{5}[0-9]{4}[A-Z][A-Z0-9]Z[A-Z0-9]$")

INSTRUCTION = (
    "Extract this invoice from the OCR text. "
    "Reply with JSON only, with keys invoice_summary and line_items."
)

def is_blank(v):
    return v is None or (isinstance(v, str) and v.strip().lower() in BLANK)

def norm_gstin(v):
    g = re.sub(r"[^A-Z0-9]", "", str(v or "").upper())
    return g if GSTIN_FULL.fullmatch(g) else ""

def valid_gstin(v):
    g = norm_gstin(v)
    return bool(g), g

def norm_field(name, v):
    if is_blank(v):
        return ""
    s = str(v).strip()
    if name in {"vendor_gstin", "customer_gstin"}:
        return norm_gstin(s)
    if name in {"tax", "total"}:
        return re.sub(r"[, ]", "", s)
    if name == "irn":
        return "".join(ch.lower() for ch in s if ch in HEX)
    return re.sub(r"\s+", " ", s.upper())

def parse_json(text):
    text = (text or "").strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except Exception:
        return None

def extract_irn_64(ocr_text):
    text = ocr_text or ""
    m = re.search(r"[a-fA-F0-9]{64}", text)
    if m:
        return m.group().lower()
    gapped = re.search(r"[a-fA-F0-9](?:[\s\-]*[a-fA-F0-9]){63}", text)
    if gapped:
        return "".join(ch.lower() for ch in gapped.group() if ch in HEX)
    return None

def find_gstins(ocr):
    compact = re.sub(r"[^A-Z0-9]", "", (ocr or "").upper())
    out, seen = [], set()
    for m in GSTIN_RE.finditer(compact):
        g = m.group()
        if g not in seen:
            seen.add(g)
            out.append(g)
    return out

def regex_fill(summary, ocr, used):
    summary = dict(summary or {})
    found = find_gstins(ocr)
    ok, _ = valid_gstin(summary.get("vendor_gstin"))
    if (not ok) and found:
        summary["vendor_gstin"] = found[0]
        used.append("regex:vendor_gstin")
    ok_c, _ = valid_gstin(summary.get("customer_gstin"))
    if (not ok_c) and len(found) >= 2:
        summary["customer_gstin"] = found[1]
        used.append("regex:customer_gstin")
    if is_blank(summary.get("irn")):
        irn = extract_irn_64(ocr)
        if irn:
            summary["irn"] = irn
            used.append("regex:irn")
    return summary

def needs_header_fallback(summary):
    summary = summary or {}
    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(summary.get(k))
            if not ok:
                return True
        elif is_blank(summary.get(k)):
            return True
    return False

def merge_header_from_v1(base_obj, v1_obj, used):
    out = deepcopy(base_obj) if isinstance(base_obj, dict) else {"invoice_summary": {}, "line_items": None}
    s = out.setdefault("invoice_summary", {})
    s1 = ((v1_obj or {}).get("invoice_summary") or {}) if isinstance(v1_obj, dict) else {}
    for k in HEADER_FIELDS:
        if k == "vendor_gstin":
            ok, _ = valid_gstin(s.get(k))
            ok1, g1 = valid_gstin(s1.get(k))
            if (not ok) and ok1:
                s[k] = g1
                used.append(f"v1:{k}")
        else:
            if is_blank(s.get(k)) and not is_blank(s1.get(k)):
                s[k] = s1[k]
                used.append(f"v1:{k}")
    out["invoice_summary"] = s
    return out

def generate(model, tokenizer, ocr_text, max_new_tokens=2048):
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr_text.strip()
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}], tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=max_new_tokens, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    return raw, parse_json(raw)

def run_gstin_repair(model, tokenizer, ocr_text, candidates):
    prompt = (
        "Assign which GSTIN is vendor and which is customer from this invoice OCR.\n"
        "Use ONLY GSTINs from the candidate list. Reply with JSON only.\n\n"
        f"Candidates: {json.dumps(candidates)}\n\nOCR text:\n{(ocr_text or '')[:8000]}"
    )
    model.set_adapter("gstin_repair")
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}], tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=128, do_sample=False, use_cache=True)
    raw = tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True).strip()
    if raw.startswith("```"):
        raw = re.sub(r"^```(?:json)?\s*", "", raw)
        raw = re.sub(r"\s*```$", "", raw)
    return parse_json(raw)

def run_full_pipeline(model, tokenizer, ocr_text):
    used = []
    model.set_adapter("v3")
    _, obj = generate(model, tokenizer, ocr_text)
    if not isinstance(obj, dict):
        obj = {"invoice_summary": {}, "line_items": None}
    summary = regex_fill(dict(obj.get("invoice_summary") or {}), ocr_text, used)

    irn = extract_irn_64(ocr_text)
    if irn:
        summary["irn"] = irn
        used.append("script:irn")
    obj["invoice_summary"] = summary

    if needs_header_fallback(summary):
        model.set_adapter("v1")
        _, obj_v1 = generate(model, tokenizer, ocr_text)
        obj = merge_header_from_v1(obj, obj_v1, used)
        summary = regex_fill(dict(obj.get("invoice_summary") or {}), ocr_text, used)
        if irn:
            summary["irn"] = irn
        obj["invoice_summary"] = summary

    cands = find_gstins(ocr_text)
    if len(cands) >= 2 and Path(GSTIN_PATH, "adapter_config.json").exists():
        repaired = run_gstin_repair(model, tokenizer, ocr_text, cands)
        if isinstance(repaired, dict):
            vg, cg = norm_gstin(repaired.get("vendor_gstin")), norm_gstin(repaired.get("customer_gstin"))
            if vg in cands and cg in cands and vg != cg:
                summary["vendor_gstin"] = vg
                summary["customer_gstin"] = cg
                used.append("gstin_repair:assign")
            else:
                used.append("gstin_repair:ungrounded")
        else:
            used.append("gstin_repair:parse_fail")
    obj["invoice_summary"] = summary
    return obj, used

def line_count(obj):
    if not isinstance(obj, dict):
        return None
    li = obj.get("line_items")
    if isinstance(li, dict) and isinstance(li.get("items"), list):
        return len(li["items"])
    if isinstance(li, list):
        return len(li)
    return None

def _count_adapter_keys(path):
    with safe_open(str(Path(path) / "adapter_model.safetensors"), framework="pt") as f:
        return len(list(f.keys()))

def _present_keys(model, adapter_name):
    return [
        k for k in model.state_dict()
        if f"lora_A.{adapter_name}.weight" in k or f"lora_B.{adapter_name}.weight" in k
    ]

def load_adapters_present(model, adapters):
    loaded, first = [], True
    for name, path in adapters:
        path = Path(path)
        assert (path / "adapter_config.json").exists(), path
        assert (path / "adapter_model.safetensors").exists(), path
        n_file = _count_adapter_keys(path)
        if first:
            model = PeftModel.from_pretrained(model, str(path), adapter_name=name)
            first = False
        else:
            if hasattr(model, "peft_config") and name in model.peft_config:
                try:
                    model.delete_adapter(name)
                except Exception:
                    pass
            model.load_adapter(str(path), adapter_name=name)
        keys = _present_keys(model, name)
        assert name in model.peft_config and len(keys) == n_file, (name, len(keys), n_file)
        print(f"PRESENT {name}: {len(keys)}/{n_file}")
        loaded.append(name)
    model.set_adapter(loaded[0])
    return model, loaded

# ---- load ----
for name in ("model", "tokenizer"):
    if name in globals():
        del globals()[name]
py_gc.collect(); torch.cuda.empty_cache()

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-1.5B-Instruct",
    max_seq_length=MAX_SEQ, dtype=None, load_in_4bit=True,
)
adapters = [("v3", V3_PATH), ("v1", V1_PATH)]
if Path(GSTIN_PATH, "adapter_config.json").exists():
    adapters.append(("gstin_repair", GSTIN_PATH))
model, loaded_names = load_adapters_present(model, adapters)
FastLanguageModel.for_inference(model)
print("loaded OK:", ", ".join(loaded_names))

# ---- score 20 ----
rows = []
with val_path.open(encoding="utf-8") as f:
    for line in f:
        if len(rows) >= LIMIT:
            break
        rec = json.loads(line)
        if rec.get("ocr_text"):
            rows.append(rec)
print(f"scoring {len(rows)} invoices (LIMIT={LIMIT})")

stats = {k: {"n": 0, "ok": 0, "missing": 0, "wrong": 0} for k in SCORE_FIELDS + ["line_count"]}
strict_ok = skipped_long = invoice_fail = scored = 0

for i, row in enumerate(rows, 1):
    ocr = (row.get("ocr_text") or "").strip()
    prompt = INSTRUCTION + "\n\nOCR text:\n" + ocr
    text = tokenizer.apply_chat_template(
        [{"role": "user", "content": prompt}], tokenize=False, add_generation_prompt=True
    )
    ntok = len(tokenizer(text, add_special_tokens=False)["input_ids"])
    if ntok > MAX_PROMPT_TOKENS:
        skipped_long += 1
        continue

    pred, used = run_full_pipeline(model, tokenizer, ocr)
    ps = dict(pred.get("invoice_summary") or {})
    gs = ((row.get("gemini_json") or {}).get("invoice_summary") or {})

    field_fail = False
    for field in SCORE_FIELDS:
        g, p = norm_field(field, gs.get(field)), norm_field(field, ps.get(field))
        stats[field]["n"] += 1
        if is_blank(ps.get(field)):
            stats[field]["missing"] += 1
            field_fail = True
        elif p == g:
            stats[field]["ok"] += 1
        else:
            stats[field]["wrong"] += 1
            field_fail = True

    gc_, pc_ = line_count(row.get("gemini_json") or {}), line_count(pred)
    stats["line_count"]["n"] += 1
    if pc_ is None:
        stats["line_count"]["missing"] += 1
        field_fail = True
    elif pc_ == gc_:
        stats["line_count"]["ok"] += 1
    else:
        stats["line_count"]["wrong"] += 1
        field_fail = True

    if all((not is_blank(ps.get(k))) and norm_field(k, ps.get(k)) == norm_field(k, gs.get(k)) for k in STRICT_FIELDS):
        strict_ok += 1
    if field_fail:
        invoice_fail += 1
    scored += 1
    print(f"scored {i}/{len(rows)}")

print("\n" + "=" * 92)
print(f"{'field':<18} {'n':>6} {'ok':>6} {'missing':>8} {'wrong':>7} {'accuracy':>10} {'fail_rate':>10}")
print("-" * 92)
total_ok = total_n = total_fail = 0
for field, s in stats.items():
    n = s["n"] or 1
    acc = 100 * s["ok"] / n
    fail = 100 * (s["missing"] + s["wrong"]) / n
    total_ok += s["ok"]; total_n += s["n"]; total_fail += s["missing"] + s["wrong"]
    print(f"{field:<18} {s['n']:>6} {s['ok']:>6} {s['missing']:>8} {s['wrong']:>7} {acc:>9.2f}% {fail:>9.2f}%")
print("-" * 92)
print(f"{'ALL FIELDS':<18} {total_n:>6} {total_ok:>6} {'':>8} {'':>7} {100*total_ok/total_n if total_n else 0:>9.2f}% {100*total_fail/total_n if total_n else 0:>9.2f}%")
print("=" * 92)
print(f"invoices scored:    {scored}")
print(f"skipped too long:   {skipped_long}")
print(f"strict-field pass:  {strict_ok}/{scored} ({100*strict_ok/scored if scored else 0:.2f}%)")
print(f"invoice fail rate:  {invoice_fail}/{scored} ({100*invoice_fail/scored if scored else 0:.2f}%)")
````
